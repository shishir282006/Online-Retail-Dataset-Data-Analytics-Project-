##### Top 5 Countries #####

```python
Country_wise_sales = df4.groupby("Country")['Total_Sale'].sum()
total_sales = Country_wise_sales.sum()

top_5_Countries = Country_wise_sales.sort_values(ascending = False).head(5)
percentage_contribution = (top_5_Countries / total_sales) * 100

plt.figure(figsize=(10,8))
bars = plt.barh(percentage_contribution.index , percentage_contribution.values)

for bar , percentage in zip(bars , percentage_contribution):
  plt.text(bar.get_width() + 1 , bar.get_y() + bar.get_height()/2 , f'{percentage:.1f}%', va = 'center' , ha = 'left')

plt.title('top 5 Countries by total sales')
plt.xlabel('percentage_contribution')
plt.ylabel('Countries')
plt.grid(axis = 'x')
plt.show()
```

##### RFM Analysis #####

```python
df4['InvoiceDate'].max()
```

```python
current_dt = df4['InvoiceDate'].max() + pd.Timedelta(days = 1)
current_dt
```

```python
rfm = df4.groupby("CustomerID").agg({
    'InvoiceDate' : lambda x : (current_dt - x.max()).days,
    'InvoiceNo' : 'count',
    'Total_Sale' : 'sum'
})
rfm.columns = ['Recency' , 'Frequency' , 'Monetary']
rfm.head(10)
```

```python
df4[df4.CustomerID == 12347]['Total_Sale'].sum()
```

```python
rfm['R_Segment']= pd.qcut(rfm['Recency'] , 4 ,labels=[4,3,2,1])
rfm['F_Segment'] = pd.qcut(rfm['Frequency'], 4 , labels = [1,2,3,4])
rfm['M_Segment'] = pd.qcut(rfm['Monetary'] , 4 , labels = [1,2,3,4])
rfm['RFM_Score'] = rfm[['R_Segment' , 'F_Segment' , 'M_Segment']].sum(axis= 1)
```

```python
rfm
```

```python
# qcut() function it will take a column and it will cut that into four quaartile and these are the labels
```

```python
rfm.drop(columns=['F_Segment'] , inplace = True)
rfm
```

```python
rfm.sort_values("RFM_Score" , ascending=False)
```

#### Customer Churn Analysis ####

```python
 df4.head(2)
```

```python
Customer_last_purchase = df4.groupby('CustomerID')['InvoiceDate'].max()
Customer_last_purchase.head()
```

```python
type(Customer_last_purchase)
```

```python
Customer_last_purchase = (current_dt - Customer_last_purchase).dt.days
Customer_last_purchase.head(10)
```

```python
# Define churn threshold(e.g , 90 days without purchase)
churn_threshold = 90
churned_customers = Customer_last_purchase[Customer_last_purchase > churn_threshold]
churned_customers.head(10)
```

```python
plt.figure(figsize=(10,6))
plt.hist(Customer_last_purchase , bins=50 , color = 'red' , alpha= 0.7)
plt.axvline(churn_threshold , color = 'black' , linestyle = 'dashed' , linewidth = 2)
plt.title("Customer Churn Distribution")
plt.xlabel("Days since last purchase")
plt.ylabel("Number of Customers")
plt.grid()
plt.show()
```
